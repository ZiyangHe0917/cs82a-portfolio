### Temperature Cutoffs

I will use these categories:
- Freezing: below 32°F
- Cold: 32°F to below 50°F
- Mild: 50°F to below 65°F
- Warm: 65°F to below 80°F
- Hot: 80°F and above

```python
temperatures = [20, 32, 45, 50, 64, 65, 75, 80, 95]

for temperature in temperatures:
    if temperature < 32:
        label = "Freezing"
    elif temperature < 50:
        label = "Cold"
    elif temperature < 65:
        label = "Mild"
    elif temperature < 80:
        label = "Warm"
    else:
        label = "Hot"

    print(temperature, "°F:", label)
```
