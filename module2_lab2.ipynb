## Step 1: Multiplication Using a `for` Loop

```python
def multiply_for(a, b):
    result = 0
    
    for i in range(b):
        result = result + a
    
    return result
```

```python
print("7 x 6 =", multiply_for(7, 6))
```

```python
print("7 x 6 =", multiply_for(7, 6))
print("5 x 4 =", multiply_for(5, 4))
print("9 x 3 =", multiply_for(9, 3))
print("8 x 2 =", multiply_for(8, 2))
```

## Step 2: Multiplication Using a `while` Loop

```python
def multiply_while(a, b):
    result = 0
    count = 0
    
    while count < b:
        result = result + a
        count = count + 1
    
    return result
```

```python
print("7 x 6 =", multiply_while(7, 6))
```

```python
print("7 x 6 =", multiply_while(7, 6))
print("5 x 4 =", multiply_while(5, 4))
print("9 x 3 =", multiply_while(9, 3))
print("8 x 2 =", multiply_while(8, 2))
```

```python
print("For loop:", multiply_for(7, 6))
print("While loop:", multiply_while(7, 6))
print("Python product:", 7 * 6)
```

## How the "While" Loop Terminates
The "WHile" Loop stops because the 'count' variable increases by 1 after each addition. When 'count' reaches 'b', the 
condition 'count<b' becomes false, so the loop ends and the function returns the accumulated result.
